# Experiment 2.1:

Write a Python program to perform basic text preprocessing on a given data.

The program should:

-> Convert text to lowercase.

-> Remove punctuation.

-> Remove numbers.  

-> Remove extra whitespace.

-> Display the cleaned text.   


```python
from google.colab import files

uploaded = files.upload()
```

```python
with open("2.1_text_data.txt", "r") as file:
  text = file.read()
```

```python
print("ORIGINAL TEXT")
print(text)
```

```python
uppercase_count = sum(1 for char in text if char.isupper())

text = text.lower()
```

```python
import string

punctuations = [char for char in text if char in string.punctuation]
different_punctuations = set(punctuations)
text = text.translate(str.maketrans("", "", string.punctuation))
```

```python
import re

numbers = [char for char in text if char.isdigit()]
number_count = len(numbers)

text = re.sub(r'\d+', '', text)
```

```python
whitespace_matches = re.findall(r'\s+', text)
extra_whitespace = sum(len(match) - 1 for match in whitespace_matches)
```

```python
text = re.sub(r'\s+', ' ', text).strip()
```

```python
print("Number of uppercase Letters:", uppercase_count)
print("Number of puntuation marks:", uppercase_count)
print("Different types of punctuation:", len(different_punctuations))
print("Punctuation types:", different_punctuations)
print("Number of digits:", number_count)
print("Amount of extra whitespace:", extra_whitespace)

print("\nCLEANED TEXT")
print(text)
```

# Experiment 2.2:

Write a Python program to perform word tokenization and sentence tokenization on a given text. Display the resulting words and sentences separately.

```python
import nltk

nltk.download('punkt')
nltk.download('punkt_tab')

from nltk.tokenize import word_tokenize, sent_tokenize


with open("2.2_tokenization_data.txt", "r") as file:
    text = file.read()

words = word_tokenize(text)

sentences = sent_tokenize(text)

print("---WORD TOKENS---")
print(words)

print("\n---SENTENCE TOKENS---")
for i, sentence in enumerate(sentences, 1):
    print(i, ":", sentence)
```

```python
import spacy

nlp = spacy.load("en_core_web_sm")

with open("2.2_tokenization_data.txt", "r") as file:
    text = file.read()

doc = nlp(text)

words = [token.text for token in doc]

sentences = [sentence.text for sentence in doc.sents]

print("------WORD TOKENS-----")
print(words)

print("\n------SENTENCE TOKENS------")
for i, sentence in enumerate(sentences, 1):
    print(i, ":", sentence)
```

```python
import re

with open("2.2_tokenization_data.txt", "r") as file:
    text = file.read()

sentences = re.split(r'(?<=[.!?])\s+', text.strip())

words = re.findall(r'\b\w+\b|[^\w\s]', text)

print("-----WORD TOKENS-----")
print(words)

print("\n-----SENTENCE TOKENS------")
for i, sentence in enumerate(sentences, 1):
    print(i, ":", sentence)
```

# Experiment 2.3:

Write a Python program to tokenize a text and remove commonly occurring stop words such as is, the, a, an, in, of, and, to, etc. Display the original and filtered tokens. (Also, display Stop Words Separately).

```python
import re

with open("2.3_clean_data.txt", "r") as file:
    text = file.read()

stop_words = {"is", "the", "a", "an", "in", "of", "and", "to", "for", "on", "with", "from", "by", "this", "that", "it"}

tokens = re.findall(r'\b\w+\b', text.lower())

stop_tokens = [word for word in tokens if word in stop_words]
filtered_tokens = [word for word in tokens if word not in stop_words]

print("Original Tokens:", tokens)
print("Stop Words:", stop_tokens)
print("Filtered Tokens:", filtered_tokens)
```
