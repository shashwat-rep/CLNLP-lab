5.1. Implement subword-level tokenization using a BPE-based tokenizer. Given
1.	Tokenize the sentence into subword units.
2.	Display the generated tokens and their corresponding token IDs.
(Implement two separate codes: a. Using pretrained model and b. Without using pretrained model)


5.1(a) BPE — Using a pretrained model

```python
with open("/content/drive/MyDrive/CNLP_LAB/CNLP-LAB/Experiment-5/input_sub_word_data.txt", "r", encoding="utf-8") as file:
    text = file.read()
```

```python
from transformers import AutoTokenizer

# Load pretrained tokenizer
tokenizer = AutoTokenizer.from_pretrained("bert-base-uncased")

# Tokenize
tokens = tokenizer.tokenize(text)

# Get token IDs
token_ids = tokenizer.convert_tokens_to_ids(tokens)

print("Input Text:")
print(text)

print("\nTokens:")
print(tokens)

print("\nToken IDs:")
print(token_ids)

print("\nToken -> ID")
for token, token_id in zip(tokens, token_ids):
    print(f"{token} -> {token_id}")
```

5.1(b) BPE — Without using a pretrained model

```python
from tokenizers import Tokenizer
from tokenizers.models import BPE
from tokenizers.trainers import BpeTrainer
from tokenizers.pre_tokenizers import Whitespace

# Read input file
with open("/content/drive/MyDrive/CNLP_LAB/CNLP-LAB/Experiment-5/input_sub_word_data.txt", "r", encoding="utf-8") as f:
    text = f.read().strip()

# Create BPE tokenizer from scratch
tokenizer = Tokenizer(BPE(unk_token="[UNK]"))

# Split input into words first
tokenizer.pre_tokenizer = Whitespace()

# Trainer
trainer = BpeTrainer(
    vocab_size=1000,
    min_frequency=1,
    special_tokens=["[UNK]", "[PAD]", "[CLS]", "[SEP]"]
)

# Train tokenizer on our input file
tokenizer.train(["/content/drive/MyDrive/CNLP_LAB/CNLP-LAB/Experiment-5/input_sub_word_data.txt"], trainer)

# Encode input
output = tokenizer.encode(text)

tokens = output.tokens
token_ids = output.ids

print("Input Text:")
print(text)

print("\nTokens:")
print(tokens)

print("\nToken IDs:")
print(token_ids)

print("\nToken -> ID")
for token, token_id in zip(tokens, token_ids):
    print(f"{token} -> {token_id}")
```

5.2(b) SentencePiece — Without using a pretrained model

```python
import sentencepiece as spm

# Train SentencePiece model
spm.SentencePieceTrainer.train(
    input="/content/drive/MyDrive/CNLP_LAB/CNLP-LAB/Experiment-5/input_sub_word_data.txt",
    model_prefix="my_sentencepiece",
    vocab_size=100,
    model_type="bpe",
    character_coverage=1.0
)

# Load our newly trained model
sp = spm.SentencePieceProcessor(
    model_file="my_sentencepiece.model"
)

# Read input
with open("/content/drive/MyDrive/CNLP_LAB/CNLP-LAB/Experiment-5/input_sub_word_data.txt", "r", encoding="utf-8") as f:
    text = f.read().strip()

# Tokenize
tokens = sp.encode(text, out_type=str)

# Token IDs
token_ids = sp.encode(text, out_type=int)

print("Input Text:")
print(text)

print("\nTokens:")
print(tokens)

print("\nToken IDs:")
print(token_ids)

print("\nToken -> ID")
for token, token_id in zip(tokens, token_ids):
    print(f"{token} -> {token_id}")
```

5.3(a) POS Tagging using spaCy

```python
import spacy

# Load English model
nlp = spacy.load("en_core_web_sm")

text = "The young student is reading an interesting book in the library."

doc = nlp(text)

# Store unique tokens
seen = set()

print("Token\tPOS\tDescription")
print("-" * 45)

for token in doc:
    if token.text not in seen:
        seen.add(token.text)

        print(
            f"{token.text}\t"
            f"{token.pos_}\t"
            f"{spacy.explain(token.pos_)}"
        )
```

5.4 POS Tagging Using spaCy + Frequency

```python
import spacy
from collections import Counter

# Load spaCy model
nlp = spacy.load("en_core_web_sm")

# Read input file
with open("/content/drive/MyDrive/CNLP_LAB/CNLP-LAB/Experiment-5/input_sub_word_data.txt", "r", encoding="utf-8") as f:
    text = f.read()

# Process text
doc = nlp(text)

# Count token frequencies
frequency = Counter(
    token.text
    for token in doc
    if not token.is_space
)

# Store information about each unique token
token_info = {}

for token in doc:

    if token.is_space:
        continue

    if token.text not in token_info:

        token_info[token.text] = (
            token.pos_,
            spacy.explain(token.pos_)
        )

print("Token\tPOS\tDescription\tFrequency")
print("-" * 80)

for token, (pos, description) in token_info.items():

    print(
        f"{token}\t"
        f"{pos}\t"
        f"{description}\t"
        f"{frequency[token]}"
    )
```
