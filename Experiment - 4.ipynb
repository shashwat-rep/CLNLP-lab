```python
import nltk
import csv
import spacy
from collections import Counter

nltk.download('punkt_tab')

nlp = spacy.load("en_core_web_sm")

# Read input text file
with open("/content/drive/MyDrive/CNLP_LAB/CNLP-LAB/Experiment-4/4.1_4.2_input.txt", "r", encoding="utf-8") as file:
    text = file.read()

# -------------------------------
# TERM FREQUENCY ANALYSIS
# -------------------------------

# Tokenize using NLTK
words = nltk.word_tokenize(text.lower())

# Keep only alphabetic words
words = [word for word in words if word.isalpha()]

# Calculate frequency
frequency = Counter(words)

# Save frequency to CSV
with open("term_frequency_4_1.csv", "w", newline="", encoding="utf-8") as file:
    writer = csv.writer(file)

    writer.writerow(["Term", "Frequency"])

    for term, freq in frequency.items():
        writer.writerow([term, freq])

# Display top 10 terms
print("Top 10 Most Frequent Terms:")

for term, freq in frequency.most_common(10):
    print(term, ":", freq)


# -------------------------------
# NAMED ENTITY RECOGNITION
# -------------------------------

doc = nlp(text)

print("\nNamed Entities:")

for entity in doc.ents:
    print(entity.text, "->", entity.label_)
```

##4.3. Write a Python program without using NLTK, spaCy, or any other NLP toolkit to:
1.	Read multiple text documents.
2.	Convert text to lowercase.
3.	Remove punctuation.
4.	Tokenize the documents.
5.	Calculate TF for every term.
6.	Calculate DF for every term.
7.	Calculate IDF using the logarithmic formula.
8.	Calculate TF-IDF for every term in every document.
9.	Display Term, TF, DF, IDF, and TF-IDF at each step.
10.	Identify the top 10 terms with the highest TF-IDF score for each document.


```python
import re
import math
from collections import Counter, defaultdict


# =========================================================
# GIVEN DOCUMENTS
# =========================================================

documents = [
    "Natural language processing is a field of artificial intelligence.",

    "Natural language processing helps computers understand human language.",

    "Machine learning is an important part of artificial intelligence."
]


# =========================================================
# STEP 1-4: LOWERCASE, REMOVE PUNCTUATION, TOKENIZATION
# =========================================================

tokenized_documents = []

print("=" * 70)
print("STEP 1-4: PREPROCESSING AND TOKENIZATION")
print("=" * 70)

for i, document in enumerate(documents):

    # Convert to lowercase
    document = document.lower()

    # Remove punctuation
    document = re.sub(r'[^\w\s]', '', document)

    # Tokenize
    words = document.split()

    tokenized_documents.append(words)

    print("\nDocument", i + 1)
    print(words)


# =========================================================
# STEP 5: CALCULATE TF
# =========================================================

term_frequencies = []

print("\n\n")
print("=" * 70)
print("STEP 5: TERM FREQUENCY (TF)")
print("=" * 70)

for i, words in enumerate(tokenized_documents):

    # Count occurrences of every term
    counts = Counter(words)

    # Total number of words
    total_words = len(words)

    tf = {}

    for term, count in counts.items():

        tf[term] = count / total_words

    term_frequencies.append(tf)

    print("\nDocument", i + 1)

    print(f"{'Term':<20}{'TF':<10}")

    for term, value in tf.items():

        print(f"{term:<20}{value:.4f}")


# =========================================================
# STEP 6: CALCULATE DF
# =========================================================

document_frequency = defaultdict(int)

for tf in term_frequencies:

    for term in tf:

        document_frequency[term] += 1


print("\n\n")
print("=" * 70)
print("STEP 6: DOCUMENT FREQUENCY (DF)")
print("=" * 70)

print(f"{'Term':<20}{'DF':<10}")

for term in sorted(document_frequency):

    print(
        f"{term:<20}"
        f"{document_frequency[term]}"
    )


# =========================================================
# STEP 7: CALCULATE IDF
# =========================================================

N = len(documents)

idf = {}

for term, df in document_frequency.items():

    idf[term] = math.log(N / df)


print("\n\n")
print("=" * 70)
print("STEP 7: INVERSE DOCUMENT FREQUENCY (IDF)")
print("=" * 70)

print(f"{'Term':<20}{'IDF':<10}")

for term in sorted(idf):

    print(
        f"{term:<20}"
        f"{idf[term]:.4f}"
    )


# =========================================================
# STEP 8: CALCULATE TF-IDF
# =========================================================

tfidf_documents = []

for tf in term_frequencies:

    tfidf = {}

    for term, tf_value in tf.items():

        tfidf[term] = tf_value * idf[term]

    tfidf_documents.append(tfidf)


# =========================================================
# STEP 9: DISPLAY TERM, TF, DF, IDF AND TF-IDF
# =========================================================

print("\n\n")
print("=" * 70)
print("STEP 9: TF-IDF")
print("=" * 70)

for i in range(N):

    print("\nDocument", i + 1)

    print(
        f"{'Term':<20}"
        f"{'TF':<10}"
        f"{'DF':<10}"
        f"{'IDF':<12}"
        f"{'TF-IDF':<12}"
    )

    print("-" * 64)

    for term in sorted(term_frequencies[i]):

        tf_value = term_frequencies[i][term]

        df_value = document_frequency[term]

        idf_value = idf[term]

        tfidf_value = tfidf_documents[i][term]

        print(
            f"{term:<20}"
            f"{tf_value:<10.4f}"
            f"{df_value:<10}"
            f"{idf_value:<12.4f}"
            f"{tfidf_value:<12.4f}"
        )


# =========================================================
# STEP 10: TOP 10 TF-IDF TERMS
# =========================================================

print("\n\n")
print("=" * 70)
print("STEP 10: TOP 10 TERMS WITH HIGHEST TF-IDF")
print("=" * 70)

for i, tfidf in enumerate(tfidf_documents):

    print("\nDocument", i + 1)

    # Sort according to TF-IDF score
    sorted_terms = sorted(
        tfidf.items(),
        key=lambda x: x[1],
        reverse=True
    )

    for term, score in sorted_terms[:10]:

        print(f"{term:<20}{score:.4f}")
```
