Experiment 3.1:  Write a Python program to perform  Write a Python program to perform stemming using the Porter Stemmer. Apply stemming to the following words. Display the original words and their stems.


```python
import nltk
from nltk.stem import PorterStemmer

stemmer = PorterStemmer()

words = [
    "playing", "played", "plays", "studies",
    "studying", "connected", "connection", "computers"
]

print(f"{'Original Word':<18}Stem")
print("-" * 30)

for word in words:
    print(f"{word:<18}{stemmer.stem(word)}")
```


Experiment 3.2:  Write a Python program to perform lemmatization on a given input words:

cats,	dogs,   running,  runs,	ran, studies, studying,	better, children, mice,	went, ate,
leaves,	caring,


```python
import nltk
from nltk.stem import WordNetLemmatizer

nltk.download("wordnet")
nltk.download("omw-1.4")

lemmatizer = WordNetLemmatizer()

words_with_pos = [
    ("cats", "n"),
    ("dogs", "n"),
    ("running", "v"),
    ("runs", "v"),
    ("ran", "v"),
    ("studies", "v"),
    ("studying", "v"),
    ("better", "a"),
    ("children", "n"),
    ("mice", "n"),
    ("went", "v"),
    ("ate", "v"),
    ("leaves", "n"),
    ("caring", "v")
]

print(f"{'Original Word':<18}Lemma")
print("-" * 30)

for word, pos in words_with_pos:
    lemma = lemmatizer.lemmatize(word, pos=pos)
    print(f"{word:<18}{lemma}")
```

Experiment 3.3: Write a Python program using Regular Expressions (Regex) to extract and process the following information from a text:
Email addresses
URLs
Mobile numbers
Hashtags
Mentions


```python
import re

text = """
Welcome to the Natural Language Processing workshop! For more information,
visit https://www.nlpworkshop.com or www.python.org. You can contact the
coordinator at nlpworkshop@gmail.com or support@python.org. For registration,
call +91-9876543210 or 9123456789. Follow us on social media @NLPWorkshop
and @PythonLearner. Share your experience using #NLP, #Python, and
#MachineLearning. You can also visit https://github.com/NLPWorkshop
for the latest updates.
"""

email_pattern = r"\b[A-Za-z0-9._%+-]+@[A-Za-z0-9.-]+\.[A-Za-z]{2,}\b"

url_pattern = (
    r"(?:https?://|www\.)"
    r"[A-Za-z0-9.-]+\.[A-Za-z]{2,}"
    r"(?:/[A-Za-z0-9._~:/?#\[\]@!$&'()*+,;=%-]*)?"
)

mobile_pattern = r"(?<!\d)(?:\+91[-\s]?)?[6-9]\d{9}(?!\d)"

hashtag_pattern = r"(?<!\w)#[A-Za-z][A-Za-z0-9_]*"

mention_pattern = r"(?<!\w)@[A-Za-z][A-Za-z0-9_]*"

emails = re.findall(email_pattern, text)
urls = re.findall(url_pattern, text)
mobile_numbers = re.findall(mobile_pattern, text)
hashtags = re.findall(hashtag_pattern, text)
mentions = re.findall(mention_pattern, text)

print("Email addresses:")
for email in emails:
    print(email)

print("\nURLs:")
for url in urls:
    print(url)

print("\nMobile numbers:")
for number in mobile_numbers:
    print(number)

print("\nHashtags:")
for hashtag in hashtags:
    print(hashtag)

print("\nMentions:")
for mention in mentions:
    print(mention)
```
